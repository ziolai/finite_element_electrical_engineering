## Section 2: Edge-Edge Interactions - Case of Intersecting Edges 

```python
# define coordinates 
x = symbols("x", real=True)
y = symbols("y", real=True)

# compute I25  
I25integrand = (x*y)/sqrt(x**2+y**2)  
I25testvalsym = float(integrate(I25integrand,(y,0,1),(x,0,1)))

# compute I26  
I26integrand = (x)/sqrt(x**2+y**2)  
I26testvalsym = float(integrate(I26integrand,(y,0,1),(x,0,1)))

# compute I27  
I27integrand = (y)/sqrt(x**2+y**2)  
I27testvalsym = float(integrate(I27integrand,(y,0,1),(x,0,1)))

# compute I28  
I28integrand = (1)/sqrt(x**2+y**2)  
I28testvalsym = float(integrate(I28integrand,(y,0,1),(x,0,1)))

# compute Q 
Qintegrand = (1-x)*(1-y)/sqrt(x**2+y**2)  
Qtestvalsym = float(integrate(Qintegrand,(y,0,1),(x,0,1)))

print("  symbolic I25 = ", I25testvalsym)
print("  symbolic I26 = ", I26testvalsym)
print("  symbolic I27 = ", I27testvalsym)
print("  symbolic I28 = ", I28testvalsym)
print("  linear comb. = ", I25testvalsym-I26testvalsym-I27testvalsym+I28testvalsym)
print("  symbolic Q   = ", Qtestvalsym)
```

## Section 3: Edge - Edge Interactions - Case of Crossing Edges    

```python
# define coordinates 
x = symbols("x", real=True)
y = symbols("y", real=True)

# compute I31  
I31integrand = (x*y)/sqrt(x**2+y**2+1)  
I31testvalsym = float(integrate(I31integrand,(y,1,2),(x,1,2)))

# compute I32  
I32integrand = (x)/sqrt(x**2+y**2+1)  
I32testvalsym = float(integrate(I32integrand,(y,1,2),(x,1,2)))

# compute I33  
I33integrand = (y)/sqrt(x**2+y**2+1)  
I33testvalsym = float(integrate(I33integrand,(y,1,2),(x,1,2)))

# compute I34  
I34integrand = (1)/sqrt(x**2+y**2+1)  
I34testvalsym = float(integrate(I34integrand,(y,1,2),(x,1,2)))

# compute Q 
Qintegrand = (1-x)*(1-y)/sqrt(x**2+y**2+1)  
Qtestvalsym = float(integrate(Qintegrand,(y,1,2),(x,1,2)))

print("  symbolic I31 = ", I31testvalsym)
print("  symbolic I32 = ", I32testvalsym)
print("  symbolic I33 = ", I33testvalsym)
print("  symbolic I34 = ", I34testvalsym)
print("  linear comb. = ", I31testvalsym-I32testvalsym-I33testvalsym+I34testvalsym)
print("  symbolic Q   = ", Qtestvalsym)
```

## Section 4: Edge - Edge Interactions - Case of Coinciding Edges - Unable to Find Primitive  

```python
# define coordinates 
x = symbols("x", real=True)
y = symbols("y", real=True)

# compute I27  
I31integrand = (x)/(sqrt(x-y)**2)  
I31testvalsym = float(integrate(I31integrand,(y,1,2),(x,1,2)))

# compute I41  
I34integrand = (x*y)/(sqrt(x-y)**2)  
I34testvalsym = float(integrate(I34integrand,(y,1,2),(x,1,2)))

# compute Q 
Qintegrand = (1-x)*(1-y)/sqrt(x**2+y**2+1)  
Qtestvalsym = float(integrate(Qintegrand,(y,1,2),(x,1,2)))

print("  symbolic I31 = ", I31testvalsym)
print("  symbolic I34 = ", I34testvalsym)
print("  linear comb. = ", I31testvalsym-I32testvalsym-I33testvalsym+I34testvalsym)
print("  symbolic Q   = ", Qtestvalsym)
```

## Section 5: Edge - Edge Interactions - Case of Parallel Edges   

```python
x = symbols("x", real=True)
y = symbols("y", real=True)
# compute I32 
I32integrand = (x)/sqrt((x-y)**2+1)  
I32testvalsym = float(integrate(I32integrand,(y,0,1),(x,0,1)))
print("  symbolic I32 = ",I32testvalsym)
# compute I47 
I47integrand = (x*y)/sqrt((x-y)**2+1)  
I47testvalsym = float(integrate(I47integrand,(y,0,1),(x,0,1)))
print("  symbolic I47 = ",I47testvalsym)
# compute Q
Qtestintegrand = ((1-x)*(1-y))/sqrt((x-y)**2+1) 
Qtestvalsym = float(integrate(Qtestintegrand,(y,0,1),(x,0,1)))
print("  symbolic Q = ",Qtestvalsym)
```
