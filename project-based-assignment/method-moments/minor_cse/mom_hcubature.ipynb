```julia
using HCubature # provides adaptive numerical integration 
using Test      # provides testing functionality 
using Plots     # provides plotting functionality

function hcubature_count(f, a, b; kws...)
    count = 0
    i = hcubature(a, b; kws...) do x
        count += 1
        # display(count)
        f(x)
    end
    return (i..., count)
end;
```
