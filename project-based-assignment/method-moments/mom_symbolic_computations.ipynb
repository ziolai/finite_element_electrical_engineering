## Section 2: One-Dimensional Poisson Equation  

Here we solve for $x \in (0,1)$ the Poisson equation 

$$
 - u''(x) = f(x)
$$

with homogeneous Dirichlet boundary conditions. 

Observe the lowering and spreading of the peak in the computed $u(x)$ due to diffusion. 

```python
from sympy import Function, dsolve, Derivative, Eq, sin, cos, symbols, pi, Lambda, Heaviside
from sympy.plotting import plot, PlotGrid
from sympy.abc import x

u, src = symbols("u src", cls=Function)
x      = symbols("x")
src    = Lambda(x, (Heaviside(x-0.45) - Heaviside(x-0.55)))

# Solve the ODE ()
eqn  = Eq(Derivative(u(x), x, x) + src(x),0)
ics0 = {u(0): 0, u(1): 0}

sol = dsolve(eqn,ics = ics0)

print(src)
print(sol)  

plotsrc = plot(src, (x, 0, 1), title="source f(x)", ylabel="f(x)", show=False)
plotu   = plot(sol.rhs, (x, 0, 1), title="solution u(x)", ylabel="u(x)", show=False)
PlotGrid(1,2,plotsrc, plotu)
```

## Section 3: Poisson Plus Zeroth Order Term 

Here we solve for $x \in (0,1)$ the Poisson equation with a zeroth-order term scaled with positive parameter $\alpha>0$ 

$$
 - u''(x) + \alpha \, u(x) = f(x)
$$

with homogeneous Dirichlet boundary conditions. Observe that the plus sign in front of the term with $\alpha$ is such that the characteristic equation of the associated homogeneous equation has real-valued roots, i.e.m, the characteristic equation is 

$$
- r^2 + \alpha = 0 \, . 
$$

Observe deformation of the tails in the computed $u(x)$ due to zeroth order term.  

```python
from sympy import Function, dsolve, Derivative, Eq, sin, cos, symbols, pi, Lambda, Heaviside
from sympy.plotting import plot, PlotGrid
from sympy.abc import x

u, src = symbols("u src", cls=Function)
x      = symbols("x")
src    = Lambda(x, (Heaviside(x-0.45) - Heaviside(x-0.55)))

# Solve the ODE ()
eqn  = Eq(u(x)*1000 - Derivative(u(x), x, x) - src(x),0)
ics0 = {u(0): 0, u(1): 0}

sol = dsolve(eqn,ics = ics0)

print(src)
print(sol)  

plotsrc = plot(src, (x, 0, 1), title="source f(x)", ylabel="f(x)", show=False)
plotu   = plot(sol.rhs, (x, 0, 1), title="solution u(x)", ylabel="u(x)", show=False)
PlotGrid(1,2,plotsrc, plotu)
```
